```python
import torch
import json
import re
from transformers import pipeline

MODEL_ID = "Qwen/Qwen2.5-1.5B-Instruct"

llm = pipeline(
    "text-generation",
    model=MODEL_ID,
    torch_dtype=torch.bfloat16 if torch.cuda.is_available() else torch.float32,
    device_map="auto",
)

print("Running on:", "GPU" if torch.cuda.is_available() else "CPU")
```

```python
def ask(system_prompt, user_message, sample=False, temperature=0.7, max_new_tokens=120):
    """Send a system prompt and a user message to the model."""

    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_message},
    ]

    kwargs = {
        "max_new_tokens": max_new_tokens,
        "return_full_text": False
    }

    if sample:
        kwargs.update(
            do_sample=True,
            temperature=temperature,
            top_p=0.9
        )
    else:
        kwargs.update(do_sample=False)

    out = llm(messages, **kwargs)[0]["generated_text"]

    if isinstance(out, list):
        out = out[-1]["content"]

    return out.strip()


# Test that the function works
print(
    ask(
        "You are a terse assistant.",
        "Say OK and nothing else."
    )
)
```

```python
LABELS = ["billing", "technical", "account_access", "other"]

TEST_SET = [
    (
        "I was charged twice for my September subscription. Please refund the extra 12,000.",
        "billing",
    ),
    (
        "The app crashes every time I try to upload a photo on Android 14.",
        "technical",
    ),
    (
        "I forgot my password and the reset email never arrives.",
        "account_access",
    ),
    (
        "Do you have an office in Abuja? Just curious.",
        "other",
    ),
    (
        "My card was declined but the money left my account and my plan is still inactive.",
        "billing",
    ),
    (
        "Video calls freeze after about two minutes on WiFi.",
        "technical",
    ),
    (
        "I want to delete my account and everything you have stored about me.",
        "account_access",
    ),
    (
        "Please change the invoice address to my company name before the next receipt.",
        "billing",
    ),
    (
        "Login says 'too many attempts' even though this is my first try today.",
        "account_access",
    ),
    (
        "Your new logo looks great, just wanted to say well done.",
        "other",
    ),
]

print(len(TEST_SET), "test cases loaded")
```

```python
SYSTEM_V1 = "You are a helpful assistant."


def prompt_v1(text):
    return f"Classify this customer message: {text}"


for text, gold in TEST_SET[:3]:
    print("GOLD:", gold)
    print("RAW :", repr(ask(SYSTEM_V1, prompt_v1(text))))
    print("-" * 70)
```

## Version 1 observations

The model generally understood the meaning of each customer message, but it did not return the required labels. It produced long explanations instead of short, machine-readable answers. The output format was inconsistent and could not be reliably processed by another program.

```python
def parse_label(raw):
    """Extract a valid label from the model's JSON response."""

    match = re.search(r"\{.*\}", raw, re.S)

    if match:
        try:
            data = json.loads(match.group(0))
            label = str(data.get("label", "")).strip().lower()

            if label in LABELS:
                return label

        except json.JSONDecodeError:
            pass

    return "UNPARSEABLE"


def evaluate(system_prompt, build_user, sample=False):
    correct = 0
    parsed = 0

    for text, gold in TEST_SET:
        raw = ask(
            system_prompt,
            build_user(text),
            sample=sample
        )

        pred = parse_label(raw)

        if pred != "UNPARSEABLE":
            parsed += 1

        if pred == gold:
            correct += 1

        flag = "PASS" if pred == gold else "FAIL"

        print(
            f"{flag} "
            f"gold={gold:<15} "
            f"pred={pred:<15} "
            f"{text[:45]}..."
        )

    n = len(TEST_SET)

    print(
        f"\nParseable: {parsed}/{n}   "
        f"Accuracy: {correct}/{n} = {correct/n:.0%}"
    )

    return correct / n
```

```python
score_v1 = evaluate(SYSTEM_V1, prompt_v1)
```

```python
SYSTEM_V2 = """You are a support-ticket routing system for a mobile fintech app.
Classify each customer message into exactly one label.

Labels:
- billing: payments, charges, refunds, invoices, pricing, failed transactions, receipts, and billing details
- technical: crashes, bugs, errors, freezing, or anything broken in the product
- account_access: login, passwords, OTP codes, lockouts, creating or deleting an account
- other: anything else, including praise, general questions, and company information

Rules:
- If a message mentions both money and access, label it by what the customer wants fixed.
- A message that only praises the product is other, never technical.
- Never invent a label outside the four labels above.
- Reply with JSON only.
- Do not provide an explanation.
- Do not use Markdown or code fences.

Output format:
{"label": "billing | technical | account_access | other", "confidence": 0.0-1.0}
"""


def prompt_v2(text):
    return f"Message: {text}"


score_v2 = evaluate(SYSTEM_V2, prompt_v2)
```

```python
SYSTEM_V2 = """You are a support-ticket routing system for a mobile fintech app.
Classify each customer message into exactly one label.

Labels:
- billing: payments, charges, refunds, invoices, pricing, failed transactions, receipts, and billing details
- technical: crashes, bugs, errors, freezing, or anything broken in the product
- account_access: login, passwords, OTP codes, lockouts, creating or deleting an account
- other: anything else, including praise, general questions, and company information

Rules:
- If a message mentions both money and access, label it by what the customer wants fixed.
- Any message about being charged, debited, refunded, invoiced, or losing money is billing, even if a service remains inactive.
- A failed or declined card transaction is billing, not technical or account_access.
- A message that only praises the product is other, never technical.
- Never invent a label outside the four labels above.
- Reply with JSON only.
- Do not provide an explanation.
- Do not use Markdown or code fences.

Output format:
{"label": "billing | technical | account_access | other", "confidence": 0.0-1.0}
"""


def prompt_v2(text):
    return f"Message: {text}"


score_v2 = evaluate(SYSTEM_V2, prompt_v2)
```

```python
FEW_SHOT = """Examples:

Message: My subscription renewed but the app still shows the Free plan.
{"label": "billing", "confidence": 0.8}

Message: The camera screen is completely black on my Samsung.
{"label": "technical", "confidence": 0.95}

Message: The OTP code never arrives so I cannot sign in.
{"label": "account_access", "confidence": 0.95}

Message: Who founded this company?
{"label": "other", "confidence": 0.9}
"""


def prompt_v3(text):
    return f"""{FEW_SHOT}

Now classify this customer message.

Message: {text}"""


score_v3 = evaluate(SYSTEM_V2, prompt_v3)
```

```python
def consistency(system_prompt, build_user, runs=3):
    stable = 0

    for text, gold in TEST_SET:
        predictions = {
            parse_label(
                ask(
                    system_prompt,
                    build_user(text),
                    sample=True
                )
            )
            for _ in range(runs)
        }

        if len(predictions) == 1:
            stable += 1
        else:
            print(
                "UNSTABLE:",
                predictions,
                "|",
                text[:50]
            )

    print(
        f"\nStable on {stable}/{len(TEST_SET)} "
        f"messages over {runs} runs"
    )

    return stable / len(TEST_SET)


consistency(SYSTEM_V2, prompt_v3)
```

```python
SYSTEM_V4 = SYSTEM_V2.replace(
    'Output format:\n{"label": "billing | technical | account_access | other", "confidence": 0.0-1.0}',
    'Think first, then answer. Keep "reason" under 15 words.\n\n'
    'Output format:\n{"reason": "...", "label": "billing | technical | account_access | other", "confidence": 0.0-1.0}'
)


score_v4 = evaluate(SYSTEM_V4, prompt_v3)
```

## Prompt Playbook

| Version | Technique used | Accuracy | Parseable | Main failure observed |
|---|---|---:|---:|---|
| v1 | Bare instruction | 0% | 0/10 | Produced long explanations instead of valid labels and JSON |
| v2 | Role, label definitions, tie-break rules, and JSON output | 90% | 10/10 | Initially confused a declined card with an inactive plan as technical |
| v3 | Version 2 plus four few-shot examples | 100% | 10/10 | No failures observed |
| v4 | Reason before label | 100% | 10/10 | No accuracy improvement; used additional tokens and processing time |

## Reflection

Adding label definitions, classification rules, and a strict JSON output contract produced the largest improvement because accuracy rose from 0% to 90%. The hardest test case was the declined card where money left the account but the plan remained inactive, because it contained both billing and technical signals. Adding a relevant billing example in Version 3 helped the model understand that payment-related problems should take priority in that situation.
